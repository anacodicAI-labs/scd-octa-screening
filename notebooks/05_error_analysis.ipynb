## 05 — Error analysis (baseline + modern)

Use saved **per-subject** predictions to find high-confidence errors and discuss **gradable** cases with the team.

- Baseline: `results/baseline/test_predictions.csv` (after `make train` with enough labels)
- Modern: `results/modern/<OPTION>/test_predictions.csv` (after `make modern_eval`)

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..").resolve()
base_csv = ROOT / "results" / "baseline" / "test_predictions.csv"
if base_csv.is_file():
    df = pd.read_csv(base_csv)
    df["err"] = (df["label"] != (df["pred_prob"] >= 0.5).astype(int))
    display(df.sort_values("pred_prob"))
else:
    print("No baseline test_predictions.csv yet — need enough labeled subjects for train/eval split.")

mod_csv = ROOT / "results" / "modern" / "optionA" / "test_predictions.csv"
if mod_csv.is_file():
    m = pd.read_csv(mod_csv)
    display(m.head(20))
```
